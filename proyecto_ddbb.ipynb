### Concatenamos los csv por vertical. df_clase_data = Data Science , df_clase_fs = Full Stack

```python
df_clase_data = pd.concat([clase_1, clase_2])
df_clase_fs= pd.concat([clase_3, clase_4])
```

### Obtenemos la tabla alumnos.

```python
alumnos_data = df_clase_data[["Nombre","Email"]].reset_index(drop=True)
alumnos_data.head(5)
```

```python
alumnos_fs = df_clase_fs[["Nombre","Email"]].reset_index(drop=True)
alumnos_fs.head(5)
```

```python
Alumnos = pd.concat([alumnos_data, alumnos_fs]).reset_index(drop=True)
Alumnos.index = Alumnos.index + 1
Alumnos = Alumnos.reset_index(names="id_alumno")
Alumnos = Alumnos.rename(columns={
    "Nombre": "nombre_alumno",
    "Email": "email"
})
Alumnos.head(5)
```

### Obtenemos la tabla Resultados_Data

```python
resultados_data = df_clase_data[["Proyecto_HLF", "Proyecto_EDA", "Proyecto_BBDD", "Proyecto_ML", "Proyecto_Deployment"]].reset_index(drop=True)
```

### Obtenemos la tabla Resultados_FS.

```python
resultados_fs = df_clase_fs[["Proyecto_WebDev", "Proyecto_FrontEnd", "Proyecto_Backend", "Proyecto_React", "Proyecto_FullSatck"]].reset_index(drop=True)
```

### Concatenamos ambas tablas Notas

```python
df_notas = pd.concat([resultados_data, resultados_fs]).reset_index(drop=True)
Notas = pd.concat([Alumnos["id_alumno"].reset_index(drop=True), df_notas], axis=1)
Notas.index = Notas.index + 1
Notas = Notas.reset_index(names="id_nota")
Notas.head(5)
```

### Obtenemos la tabla Proyectos_Data.

```python
proyecto_data = pd.DataFrame(df_clase_data.columns[5:], columns=["nombre_proyecto"]).reset_index(drop=True)
proyecto_data
```

### Obtenemos la tabla Proyectos_FS.

```python
proyecto_fs = pd.DataFrame(df_clase_fs.columns[5:], columns=["nombre_proyecto"]).reset_index(drop=True)
proyecto_fs
```

### Concatenamos y obtenemos la tabla proyectos

```python
proyectos_data = pd.DataFrame(resultados_data.columns, columns=["nombre_proyecto"])
proyectos_fs = pd.DataFrame(resultados_fs.columns, columns=["nombre_proyecto"])

Proyectos = pd.concat([proyectos_data, proyectos_fs]).reset_index(drop=True)
Proyectos.index = Proyectos.index + 1
Proyectos = Proyectos.reset_index(names="id_proyecto")

Proyectos
```
