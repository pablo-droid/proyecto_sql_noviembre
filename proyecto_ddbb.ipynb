### Concatenamos los csv por vertical. df_clase_data = Data Science , df_clase_fs = Full Stack

```python
df_clase_data = pd.concat([clase_1, clase_2])
df_clase_fs= pd.concat([clase_3, clase_4])
df_clase_fs = df_clase_fs.rename(columns={"Proyecto_FullSatck": "Proyecto_FullStack"})
```

### Obtenemos la tabla campus

```python
Campus = pd.DataFrame(
    pd.concat([df_clase_data["Campus"], df_clase_fs["Campus"]]).unique(),
    columns=["ciudad"]
)
Campus.index = Campus.index + 1
Campus = Campus.reset_index(names="id_campus")
Campus
```

### obtenemos la tabla vertical

```python
Verticales = pd.DataFrame(["Data Science", "Full Stack"], columns=["nombre_curso"])
Verticales.index = Verticales.index + 1
Verticales = Verticales.reset_index(names="id_vertical")
Verticales
```

### Concatenamos y obtenemos Promociones

```python
df_promocion_data = df_clase_data[["Promoción","Fecha_comienzo"]].reset_index(drop=True)
df_promocion_data.head(5)
```

```python
df_promocion_fs = df_clase_fs[["Promoción","Fecha_comienzo"]].reset_index(drop=True)
df_promocion_fs.head(5)
```

```python
df_promocion_data = df_clase_data[["Promoción", "Fecha_comienzo", "Campus"]].reset_index(drop=True)
df_promocion_data["nombre_curso"] = "Data Science"

df_promocion_fs = df_clase_fs[["Promoción", "Fecha_comienzo", "Campus"]].reset_index(drop=True)
df_promocion_fs["nombre_curso"] = "Full Stack"

Promociones = pd.concat([df_promocion_data, df_promocion_fs], ignore_index=True).rename(columns={
    "Promoción": "nombre_promocion",
    "Fecha_comienzo": "fecha_inicio",
    "Campus": "ciudad"
})

Promociones = (
    Promociones
    .merge(Campus, on="ciudad", how="left")
    .merge(Verticales, on="nombre_curso", how="left")
)

Promociones = (
    Promociones[["nombre_promocion", "fecha_inicio", "id_campus", "id_vertical"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
Promociones.index = Promociones.index + 1
Promociones = Promociones.reset_index(names="id_promocion")
Promociones
```

### Obtenemos la tabla alumnos.

```python
alumnos_data = df_clase_data[["Nombre","Email"]].reset_index(drop=True)
alumnos_data.head(5)
```

```python
alumnos_fs = df_clase_fs[["Nombre","Email"]].reset_index(drop=True)
alumnos_fs.head(5)
```

```python
Alumnos = pd.concat([alumnos_data, alumnos_fs], ignore_index=True)
Alumnos.index = Alumnos.index + 1
Alumnos = Alumnos.reset_index(names="id_alumno")
Alumnos = Alumnos.rename(columns={
    "Nombre": "nombre_alumno",
    "Email": "email"
})

promo_data = df_clase_data[["Promoción", "Fecha_comienzo", "Campus"]].reset_index(drop=True)
promo_data["nombre_curso"] = "Data Science"

promo_fs = df_clase_fs[["Promoción", "Fecha_comienzo", "Campus"]].reset_index(drop=True)
promo_fs["nombre_curso"] = "Full Stack"

promo_alumnos = pd.concat([promo_data, promo_fs], ignore_index=True).rename(columns={
    "Promoción": "nombre_promocion",
    "Fecha_comienzo": "fecha_inicio",
    "Campus": "ciudad"
})

promo_alumnos = (
    promo_alumnos
    .merge(Campus, on="ciudad", how="left")
    .merge(Verticales, on="nombre_curso", how="left")
    .merge(Promociones, on=["nombre_promocion", "fecha_inicio", "id_campus", "id_vertical"], how="left")
)

Alumnos = pd.concat([Alumnos, promo_alumnos[["id_promocion"]].reset_index(drop=True)], axis=1)
Alumnos.head(5)
```

### Obtenemos la tabla Proyectos_Data.

```python
proyecto_data = pd.DataFrame(df_clase_data.columns[5:], columns=["nombre_proyecto"]).reset_index(drop=True)
proyecto_data
```

### Obtenemos la tabla Proyectos_FS.

```python
proyecto_fs = pd.DataFrame(df_clase_fs.columns[5:], columns=["nombre_proyecto"]).reset_index(drop=True)
proyecto_fs
```

### Concatenamos y obtenemos Proyectos



```python
proyectos_data = pd.DataFrame(df_clase_data.columns[5:], columns=["nombre_proyecto"])
proyectos_data["nombre_curso"] = "Data Science"

proyectos_fs = pd.DataFrame(df_clase_fs.columns[5:], columns=["nombre_proyecto"])
proyectos_fs["nombre_curso"] = "Full Stack"

Proyectos = pd.concat([proyectos_data, proyectos_fs], ignore_index=True)
Proyectos = Proyectos.merge(Verticales, on="nombre_curso", how="left")
Proyectos = Proyectos[["nombre_proyecto", "id_vertical"]]
Proyectos.index = Proyectos.index + 1
Proyectos = Proyectos.reset_index(names="id_proyecto")
Proyectos
```

### Obtenemos la tabla Profesores

```python
Profesores = pd.DataFrame(claustro["Nombre"].unique(), columns=["nombre_profesor"])
Profesores.index = Profesores.index + 1
Profesores = Profesores.reset_index(names="id_profesor")
Profesores
```

### Obtenemos la tabla Modalidades

```python
Modalidades = pd.DataFrame(claustro["Modalidad"].unique(), columns=["tipo_modalidad"])
Modalidades.index = Modalidades.index + 1
Modalidades = Modalidades.reset_index(names="id_modalidad")
Modalidades
```

### Obtenemos la tabla Asignaciones_Claustro

```python
asig = claustro.rename(columns={
    "Nombre": "nombre_profesor",
    "Modalidad": "tipo_modalidad",
    "Campus": "ciudad",
    "Vertical": "nombre_curso",
    "Promocion": "nombre_promocion",
    "Rol": "rol"
})

asig.loc[asig["nombre_curso"] == "DS", "nombre_curso"] = "Data Science"
asig.loc[asig["nombre_curso"] == "FS", "nombre_curso"] = "Full Stack"

asig = asig.merge(Profesores, on="nombre_profesor", how="left")
asig = asig.merge(Modalidades, on="tipo_modalidad", how="left")
asig = asig.merge(Campus, on="ciudad", how="left")
asig = asig.merge(Verticales, on="nombre_curso", how="left")
asig = asig.merge(Promociones, on=["nombre_promocion", "id_campus", "id_vertical"], how="left")

Asignaciones_Claustro = asig[["id_profesor", "id_promocion", "id_modalidad", "rol"]]
Asignaciones_Claustro.index = Asignaciones_Claustro.index + 1
Asignaciones_Claustro = Asignaciones_Claustro.reset_index(names="id_asignacion")
Asignaciones_Claustro["id_promocion"] = Asignaciones_Claustro["id_promocion"].astype("Int64")
Asignaciones_Claustro
```

### Obtenemos la tabla Resultados_Data

```python
resultados_data = df_clase_data[["Proyecto_HLF", "Proyecto_EDA", "Proyecto_BBDD", "Proyecto_ML", "Proyecto_Deployment"]].reset_index(drop=True)
```

### Obtenemos la tabla Resultados_FS.

```python
resultados_fs = df_clase_fs[["Proyecto_WebDev", "Proyecto_FrontEnd", "Proyecto_Backend", "Proyecto_React", "Proyecto_FullStack"]].reset_index(drop=True)
```

### Concatenamos ambas tablas Notas

```python
df_notas = pd.concat([resultados_data, resultados_fs], ignore_index=True)
df_notas.insert(0, "id_alumno", Alumnos["id_alumno"])

Notas = df_notas.melt(id_vars="id_alumno", var_name="nombre_proyecto", value_name="resultado")
Notas = Notas.dropna(subset=["resultado"])
Notas = Notas.merge(Proyectos, on="nombre_proyecto", how="left")

Notas = Notas[["id_alumno", "id_proyecto", "resultado"]]
Notas = Notas.sort_values(["id_alumno", "id_proyecto"]).reset_index(drop=True)
Notas.index = Notas.index + 1
Notas = Notas.reset_index(names="id_notas")
Notas.head(10)
```
